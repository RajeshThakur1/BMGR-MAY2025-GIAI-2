## Qdrant Vector Database

Qdrant is a production-ready vector database running as a Docker container.
Unlike FAISS (in-memory) or ChromaDB (local files), Qdrant provides:
- A web dashboard at http://localhost:6333/dashboard
- REST & gRPC APIs
- Persistent storage across restarts
- Filtering, payload storage, and scalability


To install the Quardent database 

docker run -p 6333:6333 -p 6334:6334 qdrant/qdrant

````python
!pip install qdrant-client langchain-qdrant
````

````python
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams
from langchain_qdrant import QdrantVectorStore

client = QdrantClient(url='http://localhost:6333')

### Verify the connection 
print(client.get_collections())
````

### Step 1: Store Document Embeddings in Qdrant

We use the same `split_texts` (34 chunks from the PDF) and `hf_embed` model
that we created earlier. `from_documents()` embeds and stores everything automatically.

````python
COLLECTION_NAME= 'voice_ai_docs'

### create vector store and upload all document chunks

qdrant_store = QdrantVectorStore.from_documents(
    documents= split_texts,
    embedding=hf_embed,
    url='http://localhost:6333',
    collection_name = COLLECTION_NAME
)

print(f"Stored {len(split_texts)} document chunks in Qdrant collection: '{COLLECTION_NAME}'")
print("Check your dashboard: http://localhost:6333/dashboard")

````

````python
# doc_loader ---- > chunking ---> vector ---> store Vector
````

### Step 2: Similarity Search

Search for the most relevant chunks from the stored documents using natural language.

````python
query = "What are the best voice AI assistants in 2025?"

results = qdrant_store.similarity_search(query, k=3)

print(f"Query: {query}\n")
for i, doc in enumerate(results, 1):
    print(f"\n--- Result {i} ---")
    print(doc.page_content[:300])
    print(f"Source: page {doc.metadata.get('page', 'N/A')}")
````

### Step 3: Similarity Search with Relevance Scores

Get a score (0 to 1) showing how closely each result matches the query.

````python
query = "What is the market size of voice AI?"
results_with_scores = qdrant_store.similarity_search_with_score(query, k=3)

print(f"Query: {query}\n")
print("=" * 60)
for i, (doc, score) in enumerate(results_with_scores, 1):
    print(f"\n--- Result {i} | Score: {score:.4f} ---")
    print(doc.page_content[:300])
````

## RAG — Retrieval-Augmented Generation with Qdrant

RAG = **Retrieve** relevant chunks from Qdrant + **Generate** an answer using an LLM.

Flow:
```
User Question
     ↓
Qdrant (retrieve top-k relevant chunks)
     ↓
LLM (generate answer using retrieved context)
     ↓
Final Answer
```

````python
from langchain_openai import ChatOpenAI
from langchain_qdrant import QdrantVectorStore
from langchain_community.embeddings import HuggingFaceEmbeddings
# from langchain.chains import RetrievalQA
from langchain_core.prompts import PromptTemplate
````

### Step 1: Set up the Retriever

The retriever queries Qdrant and fetches the top-k most relevant chunks for a given question.

````python
COLLECTION_NAME = "voice_ai_docs"

# Reconnect to existing Qdrant collection
qdrant_store = QdrantVectorStore.from_existing_collection(
    embedding=hf_embed,
    url="http://localhost:6333",
    collection_name=COLLECTION_NAME,
)


# Convert to retriever — fetch top 4 chunks per query
retriever = qdrant_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 4}
)


# Test the retriever
test_docs = retriever.invoke("What is Alexa Plus?")

print(f"Retrieved {len(test_docs)} chunks")
print(test_docs[0].page_content[:200])
````

### Step 2: Define the Prompt Template

The prompt instructs the LLM to answer only from the provided context,
preventing hallucination.

````python
#Augumentation

RAG_PROMPT_TEMPLATE = """
You are a helpful assistant. Use ONLY the context below to answer the question.
If the answer is not in the context, say "I don't know based on the provided documents."

Context:
{context}

Question: {question}

Answer:"""

rag_prompt = PromptTemplate(
    template=RAG_PROMPT_TEMPLATE,
    input_variables=['context', 'question']
)






````

````python
# llm = ChatOpenAI(
#     model_name="gpt-4o-mini",
#     temperature=0,       # 0 = deterministic, factual answers
# )


from litellm import completion, embedding


# llm = completion(
#     model="ollama/qwen3.5:2b",
#     api_base="http://localhost:114344",
#     temperature=0
#     )


from langchain_community.chat_models import ChatLiteLLM

llm = ChatLiteLLM(
    model="ollama/qwen3.5:2b",
    temperature=0,
    api_base="http://localhost:11434",
)

````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# Assume: vectorstore and llm are already set up
retriever = qdrant_store.as_retriever()

prompt = ChatPromptTemplate.from_template("""
Answer the question based only on the following context:
{context}

Question: {question}
""")

rag_chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

response = rag_chain.invoke("what is Voice AI")
````
