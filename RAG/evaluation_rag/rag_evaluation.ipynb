### Load the knowledge base

```python
from langchain.document_loaders import DirectoryLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from dotenv import load_dotenv
load_dotenv()
loader = DirectoryLoader("./paul_graham/")

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1024, chunk_overlap=20)

documents = loader.load_and_split(text_splitter)
```

```python
documents[0].to_json()
```

```python
import os
from langchain_community.chat_models import ChatLiteLLM
from langchain_core.embeddings import Embeddings
import litellm

assert os.environ.get("OPENAI_API_KEY"), "Set the OPENAI_API_KEY environment variable before running this notebook"

# Swapped local Ollama models for OpenAI (via LiteLLM) since Ollama generation was too slow for this demo
generator_llm = ChatLiteLLM(model="gpt-4o-mini")
critic_llm = ChatLiteLLM(model="gpt-4o")


class LiteLLMEmbeddings(Embeddings):
    """Minimal langchain Embeddings adapter backed by litellm, for ragas' TestsetGenerator."""

    def __init__(self, model="text-embedding-3-small"):
        self.model = model

    def embed_documents(self, texts):
        response = litellm.embedding(model=self.model, input=texts)
        # litellm returns .data items as plain dicts, not objects with attribute access
        return [item["embedding"] for item in response.data]

    def embed_query(self, text):
        return self.embed_documents([text])[0]


ollama_emb = LiteLLMEmbeddings(model="text-embedding-3-small")
```

```python
from ragas.testset.generator import TestsetGenerator
from ragas.testset.evolutions import simple, reasoning, multi_context

generator = TestsetGenerator.from_langchain(
    generator_llm=generator_llm,
    critic_llm=critic_llm,
    embeddings=ollama_emb
)

distribution = {simple: 0.5, reasoning: 0.25, multi_context: 0.25}
testset = generator.generate_with_langchain_docs(documents,
                                                 test_size=3,
                                                 distributions=distribution,
                                                 raise_exceptions=False)
```

```python
test_df = testset.to_pandas().dropna()
```

```python
import pandas as pd

test_df = pd.read_csv('test_data_paul_graham_new.csv').dropna()
```

Next, we define a function that will accept the query engine and a question, and return the answer along with the context it looked at to generate the corresponding answer.



```python
def generate_response(query_engine, question):
    response = query_engine.query(question)
    return {
        "answer": response.response,
        "contexts": [c.node.get_content() for c in response.source_nodes],
    }
```

```python
from llama_index.core import Document as LlamaDocument, VectorStoreIndex, StorageContext, Settings
from llama_index.llms.litellm import LiteLLM
from llama_index.embeddings.litellm import LiteLLMEmbedding
from llama_index.vector_stores.qdrant import QdrantVectorStore
from qdrant_client import QdrantClient

# OpenAI (via LiteLLM) instead of local Ollama models for speed
Settings.llm = LiteLLM(model="gpt-4o-mini")
Settings.embed_model = LiteLLMEmbedding(model_name="text-embedding-3-small")

# Reuse the langchain documents/chunks already loaded above, converted to LlamaIndex's Document type
llama_documents = [
    LlamaDocument(text=doc.page_content, metadata=doc.metadata) for doc in documents
]

qdrant_client = QdrantClient(host="localhost", port=6333)
# New collection name: the old "paul_graham_essays" collection holds 1024-dim vectors from the
# previous Ollama embedding model, which doesn't match text-embedding-3-small's 1536 dims.
vector_store = QdrantVectorStore(client=qdrant_client, collection_name="paul_graham_essays_openai")
storage_context = StorageContext.from_defaults(vector_store=vector_store)

index = VectorStoreIndex.from_documents(
    llama_documents,
    storage_context=storage_context,
)

query_engine = index.as_query_engine()
```

query_engine is the query engine we defined earlier.
question is the current question, which will be picked from the CSV.
The query_engine generates a response object, from which we extract the final answer and the context and return it.

Next, we invoke the above function on every element of the test_df DataFrame, and build a new DataFrame out of it:

```python
from datasets import Dataset
from tqdm.auto import tqdm

test_questions = test_df["question"].values

responses = [generate_response(query_engine, q) for q in tqdm(test_questions)]

dataset_dict = {
    "question": test_questions,
    "answer": [response["answer"] for response in responses],
    "contexts": [response["contexts"] for response in responses],
    "ground_truth": test_df["ground_truth"].values.tolist(),
}

ragas_eval_dataset = Dataset.from_dict(dataset_dict)
```

Now, we can again use Ragas to compute the metrics we discussed earlier. Recall that in this case, we need a Critic LLM along with an embedding model to compute similarities when needed.

```python
from ragas import evaluate
from ragas.metrics import (
    faithfulness,
    answer_correctness,
    context_recall,
    context_precision,
)
```

First, we import functionalities to load an embedding model and an LLM.

Next, we import the evaluate function from the ragas library, which is used to evaluate the performance of a RAG system by applying various metrics to measure its effectiveness.

Finally, we import specific evaluation metrics from ragas.metrics which we also discussed in an earlier section.

```python
metrics = [faithfulness, answer_correctness,
           context_recall, context_precision]

# OpenAI (via LiteLLM) instead of local Ollama models for speed
critic_llm = ChatLiteLLM(model="gpt-4o")

ollama_emb = LiteLLMEmbeddings(model="text-embedding-3-small")

evaluation_result = evaluate(
    llm=critic_llm,
    embeddings=ollama_emb,
    dataset=ragas_eval_dataset,
    metrics=metrics
)
```

```python
eval_scores_df = pd.DataFrame(evaluation_result.scores)
```
