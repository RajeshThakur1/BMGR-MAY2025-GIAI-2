```python
import os
import autogen
from dotenv import load_dotenv
load_dotenv()
```

```python
### Step 1: Define the customer and support Agent
## Creating a customer agent that represent a user seeking for support

customer_agent = autogen.UserProxyAgent(
    name="Customer",
    human_input_mode="ALWAYS",
    code_execution_config={"use_docker": False},
    max_consecutive_auto_reply=5
)
```

```python
## AI Assistant Agent which will help to user's question
support_agent = autogen.AssistantAgent(
    name="SupportAgent",
    llm_config = {
        "config_list": [
            {
                "model": "gpt-3.5-turbo",
                "temperature": 0.2,
                "max_tokens": 2048,
                "api_key": os.getenv("OPENAI_API_KEY")
            }
        ]
    },
    system_message="You are a helpful and precise assistant for customer support. Always try to solve the customer's problem in one reply. If you need more information, ask the customer specific questions to get the information you need to solve the problem.",
    code_execution_config={"use_docker": False},
    max_consecutive_auto_reply=5

)
```

```python
## Step-3: Running a Simulated Customer Intrection

customer_agent.initiate_chat(support_agent, message="I need help tracing my order.")
```
